### dunder methods in python

```python
class Car():
    def __init__(self, windows, doors, enginetype):
        self.windows = windows
        self.doors = doors
        self.enginetype = enginetype

    def drive(self):
        print("The person drives the car.")

    
c = Car(4, 4, "Diesel")
print(c)
print(c.windows)
print(c.doors)
print(c.enginetype)
c.drive()
```
