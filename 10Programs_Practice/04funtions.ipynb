```python
def passport_check(pp_num):
    if len(pp_num) == 8:
        if pp_num[0] >= 'A' and pp_num[0]<= 'Z':
            status = "Valid Passport"
        else:
            status = "Invalid Passport"

    else:
        status = "Invalid Passport"
    return status

print(passport_check("M1538730"))
print(passport_check("M153830"))
print(passport_check("Z1538730"))
print(passport_check("A15730"))
```

```python
num = int(input("Please enter the number :"))
# 123
def find_sum(num):
    digit_sum = 0
    while num > 0:
        digit_sum += num % 10
        num = num // 10
    return digit_sum
    
print(find_sum(num))
```

```python
# 123
def calc_sum(num):
    stringnum = str(num)
    digit_sum = 0
    for i in stringnum:
        digit_sum = digit_sum + int(i)
    return digit_sum

calc_sum(1234)
```

```python
# *args : it will take as manay elements as we pass and then return them and use as 
def func(arg1, arg2, *arg3):
    return arg1 + arg2 + sum(arg3)

print(func(10, 12, 12, 12, 15))
```

```python
# *Kwargs
def print_three_things(**kwargs):
    print(f"First : {kwargs['first']}")
    print(f"Second : {kwargs['second']}")
    print(f"Third : {kwargs['third']}")

arguments = {'first' : 'Ice Cream', 'second' : 'Pizza', 'third' : 'Burger'}
print_three_things(**arguments)
```

```python
# Write a Python function factorial(num) which returns the factorial of a given number.

def factorial(num):
    fact = 1
    while num >= 1:
        fact = fact * num
        num = num - 1
    return fact 

print(factorial(5))
```

```python
def factorial_loop(num):
    fact = 1
    for i in range(1, num+1):
        fact = fact * i
    return fact
print(factorial_loop(5))
```

```python
# Write a Python Function is_palindrome(num) that accepts an integer num as argument and returns True if the num is palindrome else returns false. Invoke the function and based on return value print the output.
# Example is : 12321

def palindrome_check(num):
    num = str(num)
    if num[::-1] == num:
        print(f"Yes the {num} is a palindrom")
    else :
        print(f"The {num} is not a palindrom")

palindrome_check(121)
palindrome_check(1234321)
palindrome_check(342)
```
