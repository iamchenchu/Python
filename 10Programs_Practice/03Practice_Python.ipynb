```python
# Wrapper Functions

import time 

def slow_add(a: int, b: int) -> int:
    """Core function: Simulates slow work"""
    time.sleep(1)
    return a + b


def timed_add(a: int, b: int) -> int:
    """Wrapper : Times how long slow_add() takes."""
    start = time.time()
    result = slow_add(a, b)
    end = time.time()
    print(f"[TIMER] show_add function took {end - start:.4f} seconds")
    return result


ans = timed_add(10, 30)
print("Output :", ans)
```

```python
from typing import List, Dict

def raw_model_predict(texts: List[str]) -> List[float]:
    """
    Core function: pretend this is a model returning raw scores.
    Example: spam probability scores.
    """
    # Fake scores (in real life, you'd call a model)
    return [0.1 if "hello" in t.lower() else 0.8 for t in texts]

def predict_with_threshold(texts: List[str], threshold: float = 0.5) -> List[Dict]:
    """
    Wrapper: converts raw scores into structured outputs with labels.
    """
    scores = raw_model_predict(texts)  # call the core model
    
    results = []
    for text, score in zip(texts, scores):
        label = "SPAM" if score >= threshold else "NOT_SPAM"
        results.append({
            "text": text,
            "score": score,
            "label": label
        })
    return results

# Run end-to-end
samples = ["Hello team", "Win money now!!!", "hello buddy"]
outputs = predict_with_threshold(samples, threshold=0.6)
for o in outputs:
    print(o)
```

```python
def decorator(func):
    def wrapper():
        print("Before calling the function")
        func()
        print("Afrer calling the function")
    return wrapper
@decorator
def greet():
    print("Hello World")


greet()
```

```python
def print_fruits(*args):
    for fruit in args:
        print(fruit)


print_fruits('banana', "Nimbu", "Dhraksha")
```

```python
def student_details(**kwargs):
    for key, value in kwargs.items():
        print(f"{key} : {value}")


student_details(name = "Aline", age = 28, city = "Richardson", country = "India")
```

#### Labda Functions :
Python lambsa functions are small, anonymous functions defining using labda keyword, ideal for simple, one expresseion operation without needing a full 'def' statement

1. Often used along with map(), reduce(), filter(), sorted() for functional programming
2. Limited to one expression, no statements like loops or complex logic 

`function_name = lambda arg1, arg2 : Expression(arg1 + arg2)`

```python
add_ten = lambda x: x + 10
print(add_ten(10))
```

```python
c = lambda a, b : a + b 
print(c(4, 10))
```

```python
numbers = [10, 7, 6, 4, 9, 2, 7]
squared = list(map(lambda x: x**2, numbers))
print(squared)
```

```python
numbers = [1, 2, 3, 4]
squared = list(map(lambda x: x**2, numbers))
print(squared)  # Output: [1, 4, 9, 16]
```

```python
ages = [5, 3, 2, 1, 2, 22]

even_ages = list(filter(lambda x: x % 2 ==0, ages))
print(even_ages)
```

```python
from functools import reduce

numbers = [3, 9, 7, 4, 10, 14]
product = reduce(lambda x, y: x*y, numbers)
print(product)
```
